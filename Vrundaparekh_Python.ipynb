2. Check whether the columns Area and Production contain any null
values. If null values are present, first examine whether these
columns contain outliers. If outliers exist, fill the null values using
an appropriate technique. Ensure that no decimal values are
allowed in the Area column.

```python
def has_outliers(series):
    q1=series.quantile(0.25)
    q3=series.quantile(0.75)
    iqr=q3-q1
    return len(series[(series<(q1-1.5*iqr))|(series > (q3 + 1.5 * iqr))]) > 0
```

```python
for col in ['Area', 'Production']:
    if data[col].isnull().any():
        if has_outliers(data[col].dropna()):
            data[col] = data[col].fillna(data[col].median())
        else:
            data[col] = data[col].fillna(data[col].mean())
data['Area']=data['Area'].round().astype(int)
```
